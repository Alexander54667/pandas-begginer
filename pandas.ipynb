* Soal 4.4

```python
# .Round(1) untuk mengambil 1 angka di belakang koma
rata_rata = df_bersih[['matematika', 'informatika', 'bahasa_inggris']].mean().round(1)
print(rata_rata)
```

* Soal 4.5

```python
df_bersih['Rata_rata'] = df_bersih[['matematika', 'informatika', 'bahasa_inggris']].mean(axis=1).round(1)
print(df_bersih['Rata_rata'])
```

```python
def cek_status(Rata_rata):
    if Rata_rata >= 75:
        return'Tuntas'
    else:
        return'Remidial'

df_bersih['Status'] = df_bersih['Rata_rata'].apply(cek_status)
```

```python
df_bersih
```

* Soal 4.6

```python
df_bersih = df_bersih.reset_index()
df_bersih.sort_values(by='Rata_rata', ascending=False)
```

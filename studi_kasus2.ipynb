Berdasarkan hasil pengelompokan di atas, saya dapat membuktikan adanya variabel pengganggu (confounding variables) secara nyata.
• Rata-rata gaji Remote yang sebelumnya terlihat sangat tinggi (mencapai gaji_tengah Rp15,00 juta) ternyata disebabkan karena seluruh data lowongan Remote (3 lowongan) menumpuk pada profil kualifikasi tertinggi, yaitu berlokasi di Jakarta dengan syarat pendidikan S1.
• Jika di bandingkan secara adil (apple-to-apple) dengan lowongan WFO yang memiliki kualifikasi setara (Jakarta dan S1), nilai gaji_tengah-nya adalah Rp8,50 juta.
Hal ini membuktikan bahwa tingginya gaji Remote pada analisis awal dipompa oleh faktor lokasi Jakarta dan tingkat pendidikan S1. Efek murni dari sistem kerja Remote itu sendiri terhadap kenaikan gaji hanyalah sekitar Rp6,50 juta (Rp15,00 jt - Rp8,50 jt), bukan sebesar yang terlihat pada visualisasi agregat awal.

-Tahap 4.1

* **Kota manakah yang memberikan rata-rata gaji tertinggi bagi lulusan SMK, dan perusahaan apa yang menawarkan gaji tertinggi untuk lulusan tersebut di kota tersebut?**

```python
df_smk = loker[loker['pendidikan_min'].str.lower() == 'smk']
r_gaji_smk_kota = df_smk.groupby('kota')['gaji_tengah'].mean().reset_index()
r_gaji_smk_kota = r_gaji_smk_kota.sort_values(by='gaji_tengah', ascending=False)
print('-- Rata-rata Gaji lulusan SMK per Kota --')
print(r_gaji_smk_kota.to_string(index=False))
```

```python
kota_tertinggi = r_gaji_smk_kota.iloc[0]['kota']
df_smk_k_top = df_smk[df_smk['kota'] == kota_tertinggi]
baris_s_smk = df_smk_k_top.loc[df_smk_k_top['gaji_tengah'].idxmax()]
print(f'-- Perusahaan dengan Gaji SMK Tertinggi di {kota_tertinggi} --')
print(f'Perusahaan : {baris_s_smk['perusahaan']}')
print(f'Posisi : {baris_s_smk['posisi']}')
print(f'Gaji Tengah : {baris_s_smk['gaji_tengah']} Juta')
```

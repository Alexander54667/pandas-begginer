```python
def cek_duplikat(df):
    duplikat_df = df[df.duplicated(keep=False)]
    index_list = duplikat_df.index.to_list()
    print (f'Kolom yang Teridentifikasi Duplikat: {duplikat_df}')
    print (f'Jumblah Duplikat Data : {df.duplicated().sum()}')
    print (f'Baris : {index_list}')
```

```python
cek_duplikat(df)
```

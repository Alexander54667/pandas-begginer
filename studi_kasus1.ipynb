-Tahap 3.4

```python
df_kantin['stan'].describe()
```

```python
total_pendapatan = df_kantin['pendapatan'].sum()
```

```python
df_total = df_kantin.groupby('stan')[['pendapatan']].sum().reset_index()
```

```python
df_total['presentase_kontribusi'] = (df_total['pendapatan'] / total_pendapatan *100).round(2)
```

```python
df_total['presentase_kontribusi'] = df_total['presentase_kontribusi'].map('{:.2f}%'.format)
```
